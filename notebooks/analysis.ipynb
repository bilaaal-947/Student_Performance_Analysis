## Task 2: Descriptive Statistics

```python
# Task 2: Descriptive Statistics
score_cols = ['math score', 'reading score', 'writing score']

print("=" * 60)
print("DESCRIPTIVE STATISTICS")
print("=" * 60)

for col in score_cols:
    print(f"\n--- {col.upper()} ---")
    print(f"Mean:   {df[col].mean():.2f}")
    print(f"Median: {df[col].median():.2f}")
    print(f"Mode:   {df[col].mode()[0]}")
    print(f"Std:    {df[col].std():.2f}")
    print(f"Min:    {df[col].min()}")
    print(f"Max:    {df[col].max()}")
```

Mean vs Median Comparison

```python
print("MEAN vs MEDIAN COMPARISON")
print("-" * 40)
for col in score_cols:
    mean = df[col].mean()
    median = df[col].median()
    diff = mean - median
    skew = "right-skewed" if diff > 0 else "left-skewed" if diff < 0 else "symmetric"
    print(f"{col}: Mean={mean:.2f}, Median={median:.2f}, Diff={diff:.2f} → {skew}")
```

GroupBy Analysis

```python
# Average scores by gender
print("AVERAGE SCORES BY GENDER")
display(df.groupby('gender')[score_cols].mean().round(2))

# Average scores by parental level of education
print("\nAVERAGE SCORES BY PARENTAL EDUCATION")
display(df.groupby('parental level of education')[score_cols].mean().round(2))
```

Task3: Data Cleaning

```python
import random

# Create a messy version of the dataset
messy_df = df.copy()

# 1. Add missing values randomly
score_cols = ['math score', 'reading score', 'writing score']
for col in score_cols:
    idx = random.sample(range(len(messy_df)), 20)
    messy_df.loc[idx, col] = np.nan

# 2. Add duplicates
messy_df = pd.concat([messy_df, messy_df.sample(10)], ignore_index=True)

# 3. Add outliers (scores above 100 or below 0)
messy_df.loc[0, 'math score'] = 200
messy_df.loc[1, 'reading score'] = -10
messy_df.loc[2, 'writing score'] = 999

# 4. Add inconsistent text
messy_df.loc[3, 'gender'] = 'MALE'
messy_df.loc[4, 'gender'] = 'Female '  # extra space

# Save it to your data folder
messy_df.to_csv("../data/messy_students.csv", index=False)
print("Messy dataset created and saved! Shape:", messy_df.shape)
```

Identify the Problems

```python
# Load the messy data
messy = pd.read_csv("../data/messy_students.csv")

print("--- Missing Values ---")
print(messy.isnull().sum())

print("\n--- Duplicate Rows ---")
print(messy.duplicated().sum())

print("\n--- Outlier Check (Scores should be 0-100) ---")
display(messy[score_cols].describe())
```

Perform the Cleaning

```python
cleaned = messy.copy()

# 1. Fix text (remove spaces and make lowercase)
cleaned['gender'] = cleaned['gender'].str.strip().str.lower()

# 2. Remove duplicates
cleaned = cleaned.drop_duplicates()

# 3. Handle outliers (replace scores outside 0-100 with the median)
for col in score_cols:
    cleaned.loc[(cleaned[col] < 0) | (cleaned[col] > 100), col] = np.nan
    cleaned[col] = cleaned[col].fillna(cleaned[col].median())

# 4. Fill remaining missing values with median
for col in score_cols:
    cleaned[col] = cleaned[col].fillna(cleaned[col].median())

# Save the final clean version
cleaned.to_csv("../data/cleaned_students.csv", index=False)
print("Data Cleaning Complete! Final Shape:", cleaned.shape)
```

 Task 4: EDA & Correlation 
 GroupBy Analysis

```python
cleaned = pd.read_csv("../data/cleaned_students.csv")

print("By Gender:")
display(cleaned.groupby('gender')[score_cols].mean().round(2))

print("\nBy Race/Ethnicity:")
display(cleaned.groupby('race/ethnicity')[score_cols].mean().round(2))

print("\nBy Parental Education:")
display(cleaned.groupby('parental level of education')[score_cols].mean().round(2))

print("\nBy Test Preparation Course:")
display(cleaned.groupby('test preparation course')[score_cols].mean().round(2))
```

Correlation Matrix and Heatmap

```python
# Correlation Matrix
corr_matrix = cleaned[score_cols].corr()
print("Correlation Matrix:")
display(corr_matrix)

# Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap of Student Scores")
plt.tight_layout()
plt.savefig("correlation_heatmap.png")
plt.show()
```

## EDA Insights (Task 4)

**1. Gender Differences:**
Female students consistently outperform male students in reading 
and writing scores. However, male students have a slight edge in 
math. This pattern is consistent across all group comparisons.

**2. Impact of Test Preparation:**
Students who completed the test preparation course scored 
significantly higher across all three subjects compared to 
those who did not. This confirms that preparation has a 
strong positive impact on performance.

**3. Parental Education Matters:**
Students whose parents hold a master's degree score the 
highest on average. There is a clear positive correlation 
between parental education level and student performance.

**4. Race/Ethnicity Patterns:**
Group E students consistently score highest across all subjects, 
while Group A scores are generally the lowest. This may reflect 
underlying socioeconomic differences.

**5. Strong Score Correlations:**
Reading and writing scores are very highly correlated (~0.95), 
meaning students who are good readers tend to be good writers. 
Math has a moderate correlation with both (~0.80).

Task 5 — Data Visualization

 All 5 Visualizations
 

```python
# Plot 1: Histogram with KDE
plt.figure(figsize=(8, 5))
sns.histplot(cleaned['math score'], kde=True, color='steelblue', bins=20)
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.savefig("plot1_histogram.png")
plt.show()
```

```python
# Plot 2: Boxplot by Gender
plt.figure(figsize=(8, 5))
sns.boxplot(x='gender', y='math score', data=cleaned, palette='Set2')
plt.title("Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.grid(True)
plt.tight_layout()
plt.savefig("plot2_boxplot.png")
plt.show()
```

```python
# Plot 3: Scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x='reading score', y='writing score', hue='gender', data=cleaned, alpha=0.6)
plt.title("Reading Score vs Writing Score (by Gender)")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.grid(True)
plt.tight_layout()
plt.savefig("plot3_scatter.png")
plt.show()
```

```python
# Plot 4: Bar chart
edu_order = cleaned.groupby('parental level of education')['math score'].mean().sort_values().index
plt.figure(figsize=(10, 5))
sns.barplot(x='parental level of education', y='math score', data=cleaned,
            order=edu_order, palette='Blues_d')
plt.title("Average Math Score by Parental Education")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Math Score")
plt.xticks(rotation=30, ha='right')
plt.grid(True)
plt.tight_layout()
plt.savefig("plot4_barchart.png")
plt.show()
```

```python
# Plot 5: Combined 2x2 Figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Student Performance – Combined Analysis", fontsize=16)

# Subplot 1
sns.histplot(cleaned['math score'], kde=True, ax=axes[0, 0], color='steelblue')
axes[0, 0].set_title("Math Score Distribution")
axes[0, 0].grid(True)

# Subplot 2
sns.boxplot(x='gender', y='math score', data=cleaned, ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title("Math Score by Gender")
axes[0, 1].grid(True)

# Subplot 3
sns.scatterplot(x='reading score', y='writing score', hue='gender',
                data=cleaned, ax=axes[1, 0], alpha=0.6)
axes[1, 0].set_title("Reading vs Writing Score")
axes[1, 0].grid(True)

# Subplot 4
sns.barplot(x='parental level of education', y='math score',
            data=cleaned, order=edu_order, ax=axes[1, 1], palette='Blues_d')
axes[1, 1].set_title("Math Score by Parental Education")
axes[1, 1].set_xticklabels(axes[1, 1].get_xticklabels(), rotation=30, ha='right')
axes[1, 1].grid(True)

plt.tight_layout()
plt.savefig("best_visualization.png", dpi=150)
plt.show()
print("Saved as best_visualization.png")
```
